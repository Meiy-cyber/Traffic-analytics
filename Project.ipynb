```python
import pymysql
from sqlalchemy import create_engine, text

server_url = "mysql+pymysql://root:Root@localhost"
server_engine = create_engine(server_url)


new_db_name = "crash_db"


try:
    with server_engine.connect() as conn:
        conn.execute(text(f"CREATE DATABASE IF NOT EXISTS {new_db_name}"))
        conn.commit()
    print(f"Database '{new_db_name}' ")
except Exception as e:
    print(f"Error creating database: {e}")
```

```python
import pandas as pd
from sqlalchemy import create_engine

database_url = f"mysql+pymysql://root:Root@localhost/{new_db_name}"
engine = create_engine(database_url)

# Connection check
try:
    with engine.connect() as conn:
        print(f"'{new_db_name}'")
except Exception as e:
    print(f"Connection Error: {e}")
```

```python

csv_file_path ="G:\VS CODE BACKUP\Traffic_CrashesData.csv"


print("Reading CSV file...")
df = pd.read_csv(csv_file_path)


table_name = "traffic_crash_data"

try:
    print("Inserting data into MySQL... This may take a moment.")
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists="replace",  # 'replace' drops & recreates table; use 'append' to add rows
        index=False,
        chunksize=1000        # Loads in batches to avoid memory overload
    )
    print(f"Data successfully inserted into table '{table_name}'!")
except Exception as e:
    print(f"Error inserting data: {e}")
```

```python
query_1 = """
    SELECT
        weather_condition,
        crash_type,
        COUNT(*) AS total_crashes
    FROM traffic_crash_data
    GROUP BY
        weather_condition,
        crash_type
    ORDER BY
        total_crashes DESC
    LIMIT 5;
"""
df_1 = pd.read_sql(query_1, engine)
display(df_1)
```

```python
query_2 = """
    SELECT 
        street_name,
        COUNT(*) AS total_injury_crashes
    FROM 
        traffic_crash_data
    WHERE 
        injuries_total > 0
    GROUP BY 
        street_name
    ORDER BY 
        total_injury_crashes DESC
    LIMIT 10;
"""
df_2 = pd.read_sql(query_2, engine)
display(df_2)
```

```python
query_3 = """
    SELECT 
        crash_type,
        AVG(injuries_total > 0) * 100 AS injury_percentage
    FROM 
        traffic_crash_data
    GROUP BY 
        crash_type;
"""
df_3 = pd.read_sql(query_3, engine)
display(df_3)
```

```python
query_4 = """
    SELECT
        DATE_FORMAT(crash_date, '%%m') AS crash_month,
        DATE_FORMAT(crash_date, '%%H') AS peak_hour,
        COUNT(*) AS total_crashes
    FROM
        traffic_crash_data
    GROUP BY
        DATE_FORMAT(crash_date, '%%m'),
        DATE_FORMAT(crash_date, '%%H')
    ORDER BY
        total_crashes DESC;
"""
df_4 = pd.read_sql(query_4, engine)
df_4.head(15)
```

```python
query_5 = """
    SELECT 
        FIRST_CRASH_TYPE,
        COUNT(*) AS total_crashes
    FROM 
        traffic_crash_data
    WHERE 
        CRASH_HOUR >= 18
    GROUP BY 
        FIRST_CRASH_TYPE
    ORDER BY 
        total_crashes DESC
    LIMIT 5;
"""
df_5 = pd.read_sql(query_5, engine)
display(df_5)
```

```python
query_6 = """
    SELECT 
        LIGHTING_CONDITION,
        AVG(INJURIES_TOTAL) AS avg_injuries
    FROM 
        traffic_crash_data
    WHERE 
        LIGHTING_CONDITION LIKE '%%DAYLIGHT%%' 
        OR LIGHTING_CONDITION LIKE '%%DARK%%'
    GROUP BY 
        LIGHTING_CONDITION;
"""
df_6 = pd.read_sql(query_6, engine)
display(df_6)
```

```python
query_7 = """
    SELECT 
        TRAFFIC_CONTROL_DEVICE,
        AVG(INJURIES_TOTAL) AS avg_injuries
    FROM 
        traffic_crash_data
    GROUP BY 
        TRAFFIC_CONTROL_DEVICE
    ORDER BY 
        avg_injuries DESC
    LIMIT 1;
"""
pd.read_sql(query_7, engine)
```

```python
query_8 = """
    SELECT 
        LATITUDE,
        LONGITUDE,
        COUNT(*) AS total_crashes
    FROM 
        traffic_crash_data
    WHERE 
        LATITUDE IS NOT NULL 
        AND LONGITUDE IS NOT NULL
    GROUP BY 
        LATITUDE, 
        LONGITUDE
    ORDER BY 
        total_crashes DESC
    LIMIT 5;
"""
pd.read_sql(query_8, engine)
```

```python
query_9 = """
    SELECT 
        STREET_NAME,
        COUNT(*) AS total_crashes,
        AVG(INJURIES_TOTAL > 0) * 100 AS injury_rate
    FROM 
        traffic_crash_data
    WHERE 
        STREET_NAME IS NOT NULL
    GROUP BY 
        STREET_NAME
    HAVING 
        total_crashes > 100
    ORDER BY 
        injury_rate DESC
    LIMIT 5;
"""
df_9 = pd.read_sql(query_9, engine)
display(df_9)
```

```python
query_10 = """
    SELECT 
        year AS CRASH_YEAR,
        FIRST_CRASH_TYPE AS most_common_crash_type,
        COUNT(*) AS total_crashes
    FROM 
        traffic_crash_data
    GROUP BY 
        CRASH_YEAR, 
        FIRST_CRASH_TYPE
    ORDER BY 
        CRASH_YEAR ASC, 
        total_crashes DESC;
"""
df_10 = pd.read_sql(query_10, engine)
display(df_10)
```

```python
query_11 = """
    SELECT 
        CRASH_DAY_OF_WEEK,
        ROUND(COUNT(*) / 24.0, 2) AS avg_crashes_per_hour
    FROM 
        traffic_crash_data
    GROUP BY 
        CRASH_DAY_OF_WEEK
    ORDER BY 
        avg_crashes_per_hour DESC
    LIMIT 1;
"""
pd.read_sql(query_11, engine)
```

```python
query_12 = """
    SELECT 
        CASE 
            WHEN CRASH_HOUR BETWEEN 6 AND 11 THEN 'Morning'
            WHEN CRASH_HOUR BETWEEN 12 AND 16 THEN 'Afternoon'
            WHEN CRASH_HOUR BETWEEN 17 AND 21 THEN 'Evening'
            ELSE 'Night'
        END AS time_bucket,
        SUM(INJURIES_TOTAL) AS total_injuries
    FROM 
        traffic_crash_data
    GROUP BY 
        time_bucket
    ORDER BY 
        total_injuries DESC;
"""
df_12 = pd.read_sql(query_12, engine)
display(df_12)
```

```python
query_13 = """
    SELECT 
        FIRST_CRASH_TYPE,
        PRIM_CONTRIBUTORY_CAUSE,
        total_crashes
    FROM (
        SELECT 
            FIRST_CRASH_TYPE,
            PRIM_CONTRIBUTORY_CAUSE,
            COUNT(*) AS total_crashes,
            ROW_NUMBER() OVER (
                PARTITION BY FIRST_CRASH_TYPE 
                ORDER BY COUNT(*) DESC
            ) AS rnk
        FROM 
            traffic_crash_data
        GROUP BY 
            FIRST_CRASH_TYPE,
            PRIM_CONTRIBUTORY_CAUSE
    ) AS ranked_crashes
    WHERE 
        rnk <= 3;
"""
df_13 = pd.read_sql(query_13, engine)
display(df_13.head(15))
```

```python
query_14 = """
    SELECT 
        year AS CRASH_YEAR,
        COUNT(*) AS total_crashes,
        LAG(COUNT(*)) OVER (ORDER BY year) AS prev_year_crashes,
        ROUND(
            (COUNT(*) - LAG(COUNT(*)) OVER (ORDER BY year)) * 100.0 / 
            LAG(COUNT(*)) OVER (ORDER BY year), 
            2
        ) AS yoy_growth_percentage
    FROM 
        traffic_crash_data
    GROUP BY 
        year
    ORDER BY 
        year;
"""
df_14 = pd.read_sql(query_14, engine)
display(df_14)
```

```python
query_15 = """
    SELECT 
        ROUND(LATITUDE, 2) AS lat_zone,
        ROUND(LONGITUDE, 2) AS long_zone,
        COUNT(*) AS total_crashes
    FROM 
        traffic_crash_data
    WHERE 
        LATITUDE IS NOT NULL 
        AND LONGITUDE IS NOT NULL
    GROUP BY 
        lat_zone, 
        long_zone
    ORDER BY 
        total_crashes DESC
    LIMIT 10;
"""
df_15 = pd.read_sql(query_15, engine)
df_15['zone'] = df_15['lat_zone'].astype(str) + ", " + df_15['long_zone'].astype(str)
display(df_15)
```
